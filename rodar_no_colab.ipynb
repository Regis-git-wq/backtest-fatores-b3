## Resultados

```python
import pandas as pd
from IPython.display import Image, display
pd.set_option('display.width', 250); pd.set_option('display.max_columns', 20)
ler = lambda f, **k: pd.read_csv(f'results/{f}', sep=';', decimal=',', **k)
m = ler('metricas.csv', index_col=0)
display(m[m.index.isin(['G3', 'Long-Short', 'Universo EW'])].round(3))
alfas = ler('regressoes_alfa.csv', index_col=0)
display(alfas.loc[['const', 'alfa a.a.', 'Mercado']].round(4))
for g in ['destaque', 'alfas', 'robustez']:
    display(Image(f'results/{g}.png', width=900))
```
